```python
import json
from easyeditor import BaseEditor
from easyeditor import ROMEHyperParams # 以 ROME 为例

# 1. 加载你批量生成的数据集
with open('full_Questions.json', 'r', encoding='utf-8') as f:
    dataset = json.load(f)

# 这里仅以第一条数据为例进行单次测试，后续你可以写 for 循环批量跑
test_case = dataset

# 2. 核心字段映射
prompts = [test_case['edit_prompt']]
target_new = [test_case['new_counterfactual_object']]
subject = [test_case['subject']]

# 3. 映射到 EasyEdit 的 Locality 和 Portability 格式
# 注意：EasyEdit 要求传入包含 'prompt' 和 'ground_truth' 的字典结构
locality_inputs = {
    'history_matters': {
        'prompt': [test_case['metrics']['history_matters']['question']],
        'ground_truth': [test_case['metrics']['history_matters']['expected_answer']]
    },
    'time_boundary': {
        'prompt': [test_case['metrics']['time_boundary_adversarial']['question']],
        'ground_truth': [test_case['metrics']['time_boundary_adversarial']['expected_answer']]
    }
}

portability_inputs = {
    'generality': {
        'prompt': [test_case['metrics']['generality']['question']],
        'ground_truth': [test_case['metrics']['generality']['expected_answer']]
    },
    'temporal_reasoning': {
        'prompt': [test_case['metrics']['temporal_reasoning']['question']],
        'ground_truth': [test_case['metrics']['temporal_reasoning']['expected_answer']]
    }
}

# 4. 初始化配置与编辑器
# 填入你刚才修改好的 yaml 配置文件路径
hparams = ROMEHyperParams.from_hparams('./hparams/ROME/qwen-7b.yaml')
editor = BaseEditor.from_hparams(hparams)

# 5. 执行单次知识编辑与自动评估
metrics, edited_model, _ = editor.edit(
    prompts=prompts,
    target_new=target_new,
    subject=subject,
    locality_inputs=locality_inputs,
    portability_inputs=portability_inputs,
    keep_original_weight=True # 测试完后恢复模型权重，保证每次测试独立，相当于 sequential_edit=False
)

# 打印评测结果
print(metrics)
```

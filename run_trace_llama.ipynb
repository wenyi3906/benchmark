```python
import os
# Set Hugging Face mirror endpoint for faster downloads in some regions
os.environ["HF_ENDPOINT"] = "https://hf-mirror.com"

from huggingface_hub import snapshot_download

target_dir = "./huggingface_cache/models/gpt-j-6B"
print(f"🚀 Starting to download model using Hugging Face mirror to {target_dir} ...")
snapshot_download(
    repo_id="EleutherAI/gpt-j-6B",
    local_dir=target_dir,
    local_dir_use_symlinks=False,
    resume_download=True, # Enable resuming for large model downloads
    ignore_patterns=["*.h5", "*.msgpack", "*.ot"]
)
print(f"✅ Model downloaded successfully via mirror to: {target_dir}")
```
